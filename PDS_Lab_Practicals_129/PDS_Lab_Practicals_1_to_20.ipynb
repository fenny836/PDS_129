## Practical 12
**Aim:** Use SciPy's stats module to compute statistical measures and perform statistical tests on datasets.

```python
import numpy as np
from scipy import stats

d1 = np.array([12, 15, 14, 10, 18, 20, 15, 16, 14, 13])
d2 = np.array([10, 12, 11, 14, 13, 15, 12, 16, 11, 15])

print("Dataset 1: ", d1)
print("Dataset 2: ", d2)

print("Descriptive Statistics:")
r = stats.describe(d1)
print("Number of observation:", r.nobs)
print("Minimum:", r.minmax[0])
print("Maximum:", r.minmax[1])
print("Mean:", r.mean)
print("Variance:", r.variance)
print("Skewness:", r.skewness)
print("Kurtosis:", r.kurtosis)
```

```python
print("Normal Distribution")
mean = np.mean(d1)
std = np.std(d1)
x = 15

print("Mean: ", mean)
print("Standard deviation: ", std)
print("PDF: ", stats.norm.pdf(x, mean, std))
print("CDF: ", stats.norm.cdf(x, mean, std))

print("Independent T-Test:")
t_stat, p_value = stats.ttest_ind(d1, d2)
print("T-statistics:", t_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("Significance difference between two datasets")
else:
    print("No Significance difference between two datasets")
```

```python
print("Pearson Correlation:")
correlation, p_value = stats.pearsonr(d1, d2)
print("Correlation Coefficient:", correlation)
print("P-value:", p_value)

if p_value < 0.05:
    print("Correlation is statistically significant.")
else:
    print("Correlation is not statistically significant.")
```

```python
from bs4 import BeautifulSoup
import requests

search_query = "laptops"

for page_no in range(1, 3):

    url = f"https://www.flipkart.com/search?q={search_query}&page={page_no}"
    print("Fetching URL:", url)

    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
                      "AppleWebKit/537.36 (KHTML, like Gecko) "
                      "Chrome/120.0.0.0 Safari/537.36"
    }

    res = requests.get(url, headers=headers)

    if res.status_code == 200:

        soup = BeautifulSoup(res.text, "html.parser")

        # Current-style product containers
        products = soup.select("div[data-id]")

        if not products:
            print("No products found on the page")
            continue

        count = 0

        for item in products:
            text = item.get_text(" ", strip=True)

            if text:
                print("Product:", text[:200])
                print("-" * 40)

                count += 1

            if count == 4:
                break

        print("Page no:", page_no)

    else:
        print("Failed to fetch page. Status code:", res.status_code)
```

```python
# Data Loading
import pandas as pd
df = pd.read_csv("Recipe Reviews and User Feedback Dataset.csv")
features = [
    "user_reputation",
    "reply_count",
    "thumbs_up",
    "thumbs_down",
    "best_score"
]
X = df[features]
y = df["stars"]
print("Dataset Shape:", df.shape)
print("Total Records:", len(df))
print("Feature Count:", X.shape[1])
print("Prediction Target: stars")
print("\nSample Records:")
print(X.head())
```

```python
# data cleaning
print("Missing Values in Dataset:")
print(df.isnull().sum())
data = df.copy()
initial_count = len(data)
data_cleaned = data.dropna().drop_duplicates().reset_index(drop=True)
print("\nInitial Records :", initial_count)
print("Cleaned Records :", len(data_cleaned))
print("Records Removed :", initial_count - len(data_cleaned))
X = data_cleaned[features]
y = data_cleaned["stars"]
```

## Practical 15
**Aim:** Compute and visualize Binomial and Poisson probability
distributions using SciPy and Matplotlib.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom, poisson

# Problem:
# An archer hits a target with probability 0.6 and shoots 10 arrows.
# Find the probability of getting exactly 6 hits.
# Then calculate and visualize the Binomial distribution.
n = 10
p = 0.6

probability = binom.pmf(6, n, p)

print("Probability of getting exactly 6 hits:", probability)

# Values of X
x = np.arange(0, 11)

# Binomial probabilities
y = binom.pmf(x, n, p)

print("Binomial Distribution")
for i in range(len(x)):
    print("Number of Hits =", x[i], "Probability =", round(y[i], 4))
```

```python
# Problem:
# A call center receives an average of 5 calls per minute.
# Find the probability of getting exactly 4 calls in one minute.
# Then calculate and visualize the Poisson distribution.
lam = 5

probability = poisson.pmf(4, lam)

print("Probability of getting exactly 4 calls:", probability)

# Values of X
x = np.arange(0, 13)

# Poisson probabilities
y = poisson.pmf(x, lam)
print("Poisson Distribution")
for i in range(len(x)):
    print("Number of Calls =", x[i], "Probability =", round(y[i], 4))
```

```python
plt.figure(figsize=(10, 4))

# Binomial plot
plt.subplot(1, 2, 1)
plt.bar(np.arange(0, 11), binom.pmf(np.arange(0, 11), 10, 0.6), color="royalblue")
plt.title("Binomial Distribution")
plt.xlabel("Number of Hits")
plt.ylabel("Probability")

# Poisson plot
plt.subplot(1, 2, 2)
plt.bar(np.arange(0, 13), poisson.pmf(np.arange(0, 13), 5), color="purple")
plt.title("Poisson Distribution")
plt.xlabel("Number of Calls")
plt.ylabel("Probability")

plt.tight_layout()
plt.show()
```

## Practical 16
**Aim:** Plot and analyze Normal and Exponential continuous probability
distributions and demonstrate the Central Limit Theorem.

### 1. Normal Distribution
Continuous probability distribution with parameters $\mu = 60$ (mean) and $\sigma = 8$ (standard deviation).

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm, expon

# Normal Distribution parameters
mu = 60
sigma = 8
x = np.linspace(mu - 4 * sigma, mu + 4 * sigma, 1000)
normal_pdf = norm.pdf(x, mu, sigma)

# Plot Normal Distribution
plt.figure(figsize=(8, 4))
plt.plot(x, normal_pdf, color="purple")
plt.title(r"Normal Distribution ($\mu=60, \sigma=8$)")
plt.xlabel("X")
plt.ylabel("Probability Density")
plt.grid(True)
plt.show()
```

### 2. Exponential Distribution
Continuous probability distribution modeling arrival/waiting times with rate parameter $\lambda = 0.4$ (mean scale $= 1/\lambda = 2.5$).

```python
# Exponential Distribution parameters
lam = 0.4
x_exp = np.linspace(0, 15, 1000)
exponential_pdf = expon.pdf(x_exp, scale=2.5)

# Plot Exponential Distribution
plt.figure(figsize=(8, 4))
plt.plot(x_exp, exponential_pdf, color="teal")
plt.title(r"Exponential Distribution ($\lambda=0.4$)")
plt.xlabel("X")
plt.ylabel("Probability Density")
plt.grid(True)
plt.show()
```

### 3. Central Limit Theorem (CLT)
Demonstrating that the distribution of sample means approaches a Normal distribution as sample size increases, regardless of the underlying population distribution.

```python
np.random.seed(44)

# Generate an exponential population
population = np.random.exponential(scale=2.5, size=10000)

# Draw 1000 random samples of size 30 and calculate their means
sample_means = [np.mean(np.random.choice(population, size=30)) for _ in range(1000)]

# Plot Central Limit Theorem
plt.figure(figsize=(8, 4))
plt.hist(sample_means, bins=30, density=True, color="darkgreen", edgecolor="black")
plt.title("Central Limit Theorem (Distribution of Sample Means)")
plt.xlabel("Sample Mean")
plt.ylabel("Density")
plt.grid(True)
plt.show()

print("Population mean:", round(np.mean(population), 2))
print("Mean of sample means:", round(np.mean(sample_means), 2))
```

## Practical 17
**Aim:** Perform hypothesis testing using one sample t-test and interpret
results based on p-value and significance level.

```python
import numpy as np
from scipy.stats import ttest_1samp

# Sample data
data = [102, 98, 105, 99, 103, 97, 101, 104, 96, 102]

# Population mean
population_mean = 100

# Significance level
alpha = 0.05

# One-sample t-test
t_stat, p_value = ttest_1samp(data, population_mean)

print("Sample Data :", data)
print("Population mean :", population_mean)
print("t-statistic :", round(t_stat, 4))
print("P-value :", round(p_value, 4))

# Hypothesis testing
if p_value < alpha:
    print("Reject H0")
    print("There is a significant difference from the population mean")
else:
    print("Fail to reject H0")
    print("There is no significant difference from the population mean")
```

```python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv("diabetic_data.csv", na_values="?")

print("Initial Shape:", df.shape)
df.head()
```

```python
# 1. Detect & Handle Missing Values
missing = df.isnull().sum()

print("\nMissing values per column:")
print(missing[missing > 0])

for col in df.select_dtypes(include="object"):
    if df[col].isnull().any():
        df[col] = df[col].fillna(df[col].mode()[0])

print("\nRemaining missing values:", df.isnull().sum().sum())
```

```python
# 2. Detect & Remove Duplicate Records
print("\nDuplicate records:", df.duplicated().sum())

df.drop_duplicates(inplace=True)

print("Shape after removing duplicates:", df.shape)
```

```python
# 3. Detect & Treat Outliers using IQR

numeric_cols = df.select_dtypes(include=np.number).columns

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = ((df[col] < lower) | (df[col] > upper)).sum()

    print(col, ":", outliers)

    # Treat outliers by capping
    df[col] = df[col].clip(lower, upper)

print("Outliers treated successfully.")
```

```python
# Treat outliers by capping
df[col] = df[col].clip(lower, upper)

print("Outliers treated successfully.")
print("Final Shape:", df.shape)
df.head()
```

## Practical 19
**Aim:** Apply data preprocessing techniques: slicing, filtering, sorting,
concatenating, aggregating and normalizing data using Pandas.

```python
import pandas as pd
import numpy as np
# Load dataset
df = pd.read_csv("diabetic_data.csv", na_values="?")
```

```python
# 1. Slicing
print("Slicing:")
print(df.loc[0:4, ["race", "gender", "age", "time_in_hospital"]])
```

```python
# 2. Filtering
print("Filtering:")
filtered = df[df["time_in_hospital"] > 5]
print(filtered[["age", "gender", "time_in_hospital"]].head())
```

```python
# 3. Sorting
print("Sorting:")
sorted_df = df.sort_values("num_medications", ascending=False)
print(sorted_df[["age", "num_medications"]].head())
```

```python
# 4. Concatenating
part1 = df[["encounter_id", "age", "gender"]].head(5)
part2 = df[["encounter_id", "age", "gender"]].tail(5)

combined = pd.concat([part1, part2])
print("Concatenated Data:")
print(combined)
```

```python
# 5. Aggregating
print("Aggregation:")
aggregation = df.groupby("gender")["time_in_hospital"].agg(["mean", "min", "max", "count"])
print(aggregation)
```

```python
# 6. Normalizing
print("Normalization:")
df["normalized_medications"] = (
    (df["num_medications"] - df["num_medications"].min()) /
    (df["num_medications"].max() - df["num_medications"].min())
)
print(df[["num_medications", "normalized_medications"]].head())
```

## Practical 20
**Aim:** Create various types of plots using Matplotlib: line plot, bar chart,
histogram, pie chart, scatter plot using subplots and figure
properties.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("diabetic_data.csv", na_values="?")

plt.figure(figsize=(12, 8))

# line plot
mean_stay = df.groupby("age")["time_in_hospital"].mean()

plt.subplot(2, 3, 1)
plt.plot(mean_stay.index, mean_stay.values, marker="o")
plt.title("Average Hospital Stay by Age")
plt.xlabel("Age Group")
plt.ylabel("Average Hospital Stay")
plt.xticks(rotation=45)

# bar chart
admission_count = df["admission_type_id"].value_counts().sort_index()

plt.subplot(2, 3, 2)
plt.bar(admission_count.index.astype(str), admission_count.values)
plt.title("Patients by Admission Type")
plt.xlabel("Admission Type ID")
plt.ylabel("Number of Patients")

# histogram
plt.subplot(2, 3, 3)
plt.hist(df["num_medications"].dropna(), bins=20)
plt.title("Distribution of Number of Medications")
plt.xlabel("Number of Medications")
plt.ylabel("Frequency")

# pie chart
readmitted = df["readmitted"].value_counts()

plt.subplot(2, 3, 4)
plt.pie(readmitted.values,
        labels=readmitted.index,
        autopct="%1.1f%%")
plt.title("Readmission Distribution")

# scatter plot
sample = df[["num_lab_procedures", "num_medications"]].dropna().head(1000)

plt.subplot(2, 3, 5)
plt.scatter(sample["num_lab_procedures"], sample["num_medications"], alpha=0.5)
plt.title("Lab Procedures vs Medications")
plt.xlabel("Number of Lab Procedures")
plt.ylabel("Number of Medications")

# Add space between rows and columns
plt.subplots_adjust(hspace=0.6, wspace=0.5)

plt.show()
```
